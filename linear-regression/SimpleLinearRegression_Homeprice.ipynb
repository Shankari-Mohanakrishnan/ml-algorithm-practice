# Simple Linear Regression - Home Price Prediction

## Objective
This project demonstrates how Simple Linear Regression can be used to predict house prices based on area.

We use:
- `Area` as the independent variable (X)
- `Price` as the dependent variable (y)

The model learns the relationship between area and house price using the equation:

\[
y = mx + b
\]

Where:
- `y` = Predicted house price
- `m` = Slope (coefficient)
- `x` = Area
- `b` = Intercept

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
%matplotlib inline
```

# Load Dataset

The dataset contains:
- Area of the house
- Corresponding house price

We load the CSV file into a Pandas DataFrame for analysis and model building.

```python
df=pd.read_csv('homeprices.csv')
df.head()
```

# Data Visualization

A scatter plot is created to visualize the relationship between:
- House Area
- House Price

This helps us check whether a linear relationship exists between the variables.

Observation:
- As the area increases, the house price also increases.
- The data appears approximately linear, making Linear Regression a suitable algorithm.

```python
plt.xlabel('area')
plt.ylabel('price')
plt.scatter(df.area, df.price,marker='*')
plt.title('Scatter plot for Area Vs Home Price')
```

# Define Independent and Dependent Variables

We separate the dataset into:

### Independent Variable (Feature)
- `X = area`

### Dependent Variable (Target)
- `y = price`

The model will learn how changes in area affect house price.

```python
#Dependent & Independent Variables 
X=df[['area']]
y=df[['price']]
```

```python
X
```

```python
y
```

# Train Linear Regression Model

We create and train the Linear Regression model using the training data.

The model learns:
- Slope (`m`)
- Intercept (`b`)

These values are used to form the best fit line.

```python
model=LinearRegression()
model.fit(X,y)
```

# Model Parameters

After training, the model calculates:

## Slope (Coefficient)
Represents how much the house price increases for every one unit increase in area.

## Intercept
Represents the predicted house price when area is 0.

The fitted linear regression equation becomes:

\[
Price = (Slope \times Area) + Intercept
\]

```python
#slope
model.coef_
```

```python
#intercept
model.intercept_
```

# House Price Prediction

We use the trained model to predict house prices for new house areas.

Examples:
- Predict price for 3500 sq ft house
- Predict price for 6000 sq ft house

The model uses the learned regression equation to estimate the prices.

```python
model.predict([[3500]])
```

```python
model.predict([[6000]])
```

# Conclusion

Simple Linear Regression successfully modeled the relationship between:
- House Area
- House Price

Key Learnings:
- Linear Regression works well when variables have a linear relationship.
- The algorithm finds the best fit line using slope and intercept.
- The trained model can predict future house prices for unseen areas.
